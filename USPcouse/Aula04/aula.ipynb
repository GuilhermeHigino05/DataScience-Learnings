**PANDAS**

É uma biblioteca que facilita a manipulação de dados tabulares
 - Conexão com diversas fontes de dados
    O pandas é capaz de ler e escrever sobre diversos tipos de dados com CSV, Excel, Parquet.
 - Estrutura de Dados expressivas
  Series e Dataframes modelas dados de forma amigável e expões metodos poderosos

*Tipos de dados no Pandas*

O Pandas utiliza por padrão a biblioteca numpy para representar suas estruturas de dados

 - Tipos de dados usuais e compostos -> int, float e bool -> Object: listas, dicionários e Strings
 - Categorias: O tipo category pode ser usado para guardar categorias que se repetem com alta frequência de forma otimizada em gastos na memória
 

*Estrutura de dados SERIES*

As series são sequências unidimencionais de dados

São compostas de:
 - Um conjuto de dados unidimencional (de qualquer tipo)
 - Um índice que é usado para acessar cada dado da serie

```python
# Criando series
import pandas as pd

serie1 = pd.Series([1, 2, 3])

serie2 = pd.Series({"A":1, "B": 2, "C": 3})

print(serie1[1])

print(serie2["B"])
```

*DataFrame(DF)*

São a classe mais utilizada no Pandas.
Representam uma tabela com várias linhas e colunas.

 - Index: Indexa as rows do DF, geralmente é numérico, mas pode ser composto por Objects
 - Colunas: Cada coluna do DF é uma *Series* com certos valores

*Leitura de Arquivos*

 - *CSV*: simples e intuitivo

```python
df = pd.read_csv(
    'path_to_file.csv', # Obrigatório, pode ser um link
    #parâmetros opcionais
    sep=';',
    #Separador do CSV, por padrão, mas pode ser especifico
    index_col='indice',
    #Qual coluna do csv se tornará o indice, por padrão nenhuma
    usecols=["col1", "col2"],
    #quais colunas ler
    encoding="latin-1",
    #Qual encoding usar para ler os caracteres do csv
)
```

 - Excel: Precisa da biblioteca auxiliar openpyxl

```python
df = pd.read_excel(
    'path_to_file.xlsx',
    sheet_name='sheet1', #obrigatório caso o arquivo tenha mais de uma sheet

    #params opcionais similares ao read_csv
    usecols=[],
    index_col='col1'
)
```

*View X copy*

Já que DFs podem usar bastante espaço na memória, a maioria das operações acontecem em uma passagem por referência do objeto

view_df = df[['col1', 'col2']] Cria uma view do DF original

view_df.iat[0, 2] = 2  Modifica o df original também

view_df = df[['col1', 'col2']].copy() Cria uma cópia do DF original

view_df.iat[0, 2] = 2  Não modifica o df original

*Merge/Join*

O merge permite o Pandas juntar 2 ou mais DataFrames

Tanto no sql quanto nos DFs do Pandas, o Merge/Join é uma operação que resulta em uma nova tabela com colunas de ambas estrutras anteriores

SINTAX

novo_df = df.esq.merge(
    df_dir,  #Dataframe a ser juntado
    how="inner", #Poder ser left, right, outer...
    on="col_comum", #Qual coluna será usada para o juntar os valores

    #Alternativas para o parâmetro on:
    left_on="col_esq",
    right_on="col_dir",
    sort=True #Ordena ou não os valores do novo DF
)
