```python
import pandas as pd
```

```python
df = pd.read_csv(
    'datasets_treino_pandas.csv'
)

df[:5]
```

```python
#function date standardization 

def standardizesDates(dataframe, column_name):
    dataframe[column_name] = pd.to_datetime(dataframe[column_name], format='mixed', dayfirst=True);
    return dataframe
```

```python
df = standardizesDates(df,'Data_Pedido')

df['Data_Pedido']
```

```python
df['Desconto'] = df['Desconto'].fillna('Desconto não aplicado')
df['Lucro'] = df['Lucro'].replace('Sem lucro', '0')
df.isna().sum()
```

```python
df['Lucro'] = pd.to_numeric(df['Lucro'], errors='coerce')
df['Valor_Venda'] = pd.to_numeric(df['Valor_Venda'], errors='coerce')

df['Custo'] = df['Valor_Venda'] - df['Lucro'];
print(df[['Custo', 'Valor_Venda', 'Lucro']]);

```

```python
#Sinalizador de rentabilidade

df['Alta_Rentabilidade'] = df['Lucro'] > 200
df['Alta_Rentabilidade']
```

```python
MaioresLucros = df.groupby('Produto')['Lucro'].sum().nlargest(5)
MaioresLucros
```
