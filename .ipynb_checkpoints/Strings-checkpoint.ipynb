```python
import re
```

```python
s = 'Corinthians melhor time do mundo';
partter = r'melhor'
d = re.search(partter, s)

if d:
    print(f'{partter} encontrado')
else:
    print(f'{partter} not found')
```

```python
n = 'the phone is 1234567890'

partter = r'\d\d\d\d\d\d\d\d\d\d'

match = re.search(partter, n)

if match:
    print(f'Numero encontrado {match.group()}')
else:
    print('no match')
```
