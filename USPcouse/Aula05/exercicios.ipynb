```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
dataset = sns.load_dataset('geyser')
df = pd.DataFrame(data=dataset)
df.head()
```

```python
#Exercicio 1

plt.scatter(df['duration'], df['waiting']);
plt.title('Relação entre tempo de espera e duração de gêiseres');
plt.xlabel('Duração');
plt.ylabel('Espera');

```

```python
#Exercico 2
#Anscombe
anscombe = sns.load_dataset('anscombe')

df2 = pd.DataFrame(data=anscombe)

fg, axs = plt.subplots(2, 2);

line_x = np.linspace(3, 20, 100);
line_y = 3 + 0.5 *line_x

fg.suptitle('Anscombe')

axs[0,0].set_title('I')
mask_I =df2['dataset'] == 'I'
axs[0,0].scatter((df2[mask_I]['x']), (df2[mask_I]['y']))
axs[0,0].plot(line_x,line_y, c='red');

axs[1,0].set_title('II')
mask_II =df2['dataset'] == 'II'
axs[1,0].scatter((df2[mask_II]['x']), (df2[mask_II]['y']))
axs[1,0].plot(line_x,line_y, c='red');

axs[0,1].set_title('III')
mask_III =df2['dataset'] == 'III'
axs[0,1].scatter((df2[mask_III]['x']), (df2[mask_III]['y']))
axs[0,1].plot(line_x,line_y, c='red');

axs[1,1].set_title('IV')
mask_IV =df2['dataset'] == 'IV'
axs[1,1].scatter((df2[mask_IV]['x']), (df2[mask_IV]['y']))
axs[1,1].plot(line_x,line_y, c='red');
```

```python
#Exercicio 3

titanic = sns.load_dataset('titanic');
df3 = pd.DataFrame(data=titanic);

fg, axs = plt.subplots(1,2, layout='tight');

fg.set_size_inches(8, 4)

axs[0].set_title('Idades de pessoas no Titanic');
axs[0].hist(df3['age'], bins=10)
axs[0].set_xlabel('Idade')
axs[0].set_ylabel('Frequência');

axs[1].set_title('Gênero de pessoas no Titanic');
gender_counts = df3['sex'].value_counts()
axs[1].bar(gender_counts.index, gender_counts.values, color=['red', 'blue']);
axs[1].set_ylabel('Frequência');
```
