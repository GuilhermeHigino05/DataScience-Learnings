Um quadrado mágico é todo quadrado de lado *n* que os inteiros de 1 a *n²* aparecem sem repetição e em que as somas dos elemetos das linhas, colunas e diagonais são iguais.

Faça uma função que receba uma matriz quadrada e qu verifique se ela é um quadrado mágico ou não.

```python
import numpy as np
```

```python
def VerificaQuadradoMagico(mtx):
    if(np.unique(mtx).size != mtx.size):
        return False
    alvo = np.sum(mtx, axis=0)[0]
    hori_ok = np.all(np.sum(mtx, axis=0) == alvo)
    vert_ok = np.all(np.sum(mtx, axis=1) == alvo)
    diagon_ok = np.sum(np.diagonal(mtx)) == alvo
    if hori_ok & vert_ok & diagon_ok:
        return True
    else: 
        return False
```

```python
print(VerificaQuadradoMagico(np.array([[2, 7, 6], [9, 5, 1], [4, 3, 8]])))
```
