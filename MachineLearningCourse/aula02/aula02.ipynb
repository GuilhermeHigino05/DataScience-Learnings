**O perceptron me da a mehor decision boundary?**

 - Queremos deixar uma margem na Decision Boundary
 - Essa margem irá servir para a generalização dos dados ainda não vistos

A fórmula para calcular a distância é $ d = \frac{1}{||\theta||} $


**Função Gamma** <br>
*Objetivos*
 - Verifica se o ponto foi classificado de forma correta <br>
   $y_i(\vec\theta \cdot \vec x + \theta_0) \le 0$  #Foi classificado de forma errada
 - Verifica se o ponto está dentro ou fora da margem <br>
   $ d(\vec x_i, \theta . \vec x + \theta_0 = 0) >  \frac{1}{||\theta||}$

**Função Loss**

Funcão Loss da última aula sem o calculo da distância $Loss = y_i(\vec\theta \cdot \vec x + \theta_0) \le 0$ <br>

A nova função Loss deve ser capaz de quantificar:
 - Erros e acertos
 - Fora ou dentro da margem 
 - Se erro a Magnitude desse erro (distância) <br>

Ela deve penalizar mais erros que estão a uma distância maior da margem

**Regularização**

 - É a técnica que nos permite a expandir a capacidade de generalização do nosso modelo
 - Fazer com que nosso modelo não fique hiper fixado no nossos dados de treinamento <br>


 Nosso novo objetivo é minimizar a $Loss$ e maximizar a margem sem perder aprendizado ou deixar nosso modelo muito simples, para isso vamos utilizar regularização do tipo __L2__ que é: $ \frac{1}{2} ||\theta||^2 $;


 Com essa função matemática definida, queremos agora o $min(Loss)and$ $min(\frac{1}{2} ||\theta||^2)$ resumindo, minimizar a função de perda e maximizar a margem(distância).

**Função Objetivo**

Queremos minimizar a seguinte função objetivo: <br>
$$ 
    J(\theta, \theta_0) =  \frac{1}{n} \sum_{i=1}^{n} Loss_h(y_i(\vec\theta \cdot \vec x + \theta_0)) + \frac{\lambda}{2} ||\theta||^2
$$

Essa função J é composta por duas partes, a primeira parte é a função Loss e a segunda parte é a função de regularização L2, que tem como objetivo maximizar a margem. Ela é chamada de empirical risk, ela nos fornece o error inteiro do modelo, e a função de regularização é chamada de regularization term, ela nos fornece a capacidade de generalização do modelo. <br>

**Quem é nosso valor de regularização $\lambda$?** <br>
O valor de $\lambda$ é um hiperparâmetro que nos permite controlar a importância da função de regularização em relação a função de perda, ou seja, quanto maior o valor de $\lambda$, mais importância a função de regularização terá, e quanto menor o valor de $\lambda$, mais importância a função de perda terá. <br>

Ele define o grau de importância que $\theta$ tem em $J$

**Lambda (λ)**

A complexidade do nosso modelo é definida por $c = \frac{1}{\lambda}$
 - Lambda alto -> baixa complexidade, modelo simples, margem grande, mas com baixo aprendizado <br>

 - Lambda baixo -> alta complexidade, modelo complexo, margem pequena, mas com alto aprendizado <br>

    - Modelo muito complexo é aquele que está super ajustado aos dados de treinamento
        - Chamamos isso de overfitting, que é quando o modelo aprende muito bem os dados de treinamento, mas não consegue generalizar para novos dados <br>
        

**Derivada** <br>

A derivada é um conceito fundamental do cálculo que representa a taxa de variação instantânea de uma função em relação a uma de suas variáveis. Em outras palavras, a derivada nos informa como a função muda à medida que a variável independente muda, ou seja, ao calcular a derivada em um ponto específico você será capaz de saber a natureza da atualização daquela função. A derivada é frequentemente representada pelo símbolo $f'(x)$ ou $\frac{df}{dx}$, onde $f$ é a função e $x$ é a variável independente. <br>

Para exemplificar iremos utilizar a função $f(x) = x^2$ e calcular sua derivada. A derivada de $f(x)$ em relação a $x$ é dada por:
$$
f'(x) = \lim_{{h \to 0}} \frac{f(x + h) - f(x)}{h}
$$

A função $f(x) = x^2$ é uma parábola que tem seu ponto minímo em $x = 0$
