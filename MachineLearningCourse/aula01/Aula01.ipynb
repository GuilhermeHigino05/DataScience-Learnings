**Introduction To Machine Learning**

Formalmente, o campo de Machine Learning busca criar modelos que, perante um ambiente (representado pelos dados numéricos), possam se adequar de modo a satisfazer um objetivo nesse meio,
via o processo de aprendizado.

Tudo isso é meio abstrato, então vamos concretizar um pouco essa introdução. Na vida real, todos
já usamos alguma vez o ChatGPT: ele é um modelo que utiliza como dados inúmeros discursos humanos, formando internamente noções lógico-linguísticas;



*1.2 Tipos de modelos*

 - Supervisionado: utilizam-se labels para tentar quantizar a qualidade de previsão do modelo. \\\
Separado em 2 tipos: Classificação (quantia discreta de possíveis previsões, na forma de classes) e Regressão (espaço de output contínuo). Ex: Regressão Linear e SVM’s.

 - Não-Supervisionado: não há labels, e o modelo deve usar algum critério interno para agrupar
os dados e diferenciá-los. Ex: K-Means e PCA.

 - Semi-Supervisionado: um campo um tanto cinza entre os dois itens anteriores, onde às vezes
utilizam-se labels, e às vezes não. É aqui onde muitos dos modelos mais promissores da atualidade se encaixam. Ex: Reinforcement Learning, Algoritmos Genéticos e Modelos Generativos.

**Classificação Linear**

 - Cada dado está relacionado a uma classe 
    - Ex: Classificar de um filme é bom ou ruim

O classificador linear vai se classificar em dois tipos de dados, dados linearmente separáveis
 - Objetivo: encontrar a função linear
 - A dimenção do vetor de característica dita será uma reta (__R²__), ou um hiperplano (__R^N__);

*Decision Boundary* <br>

 Em um classificador binário, a decision boundary é a reta (R²), plano (R³) ou hiperplano (R^n, n > 3) que separa as partes.

*Vetor de parâmetros*<br>

 - O velor de parâmetros descreve a decision boundary<br>

Feature vector Genérico $\vec{x}$ = {$x_1$,$x_2$} <br>

__Exemplo:__ vetor de parâmetros $\vec{\theta}$ = {$1$,$1$} <br>

isso define a reta: 1.$x_1$ + 1.$x_2$

 - Podemos descreve-la como: $\vec{\theta}$ . $\vec{x}$ = $0$





O ${\theta}$ é nosso vetor de parâmetros (dados na qual nosso modelo foi treinado) e o ${x}$ é a variável que queremos classificar.

Portanto se $\vec{\theta}$ . $\vec{x}$ > $0$ as bolas são azuis <br>
Exemplo: $\vec{x}$ = {1,2}, $\vec{\theta}$ = {1,1}:  $\vec{\theta}$ . $\vec{x}$ = $3$

E se $\vec{\theta}$ . $\vec{x}$ < $0$ as bolas são vermelhas <br>
Exemplo: $\vec{x}$ = {-2,1}, $\vec{\theta}$ = {1,1}:  $\vec{\theta}$ . $\vec{x}$ = $-1$


**Perceptron**

   - Inicializar o vetor de parâmetros como 0: ${\theta}$ = $0$.

   - Iterar sobre os dados fazendo operações ajustando o $\vec{\theta}$ sempre que o resultado for mal classificado.

   - No final das iterações teremsos construído a decision boundary.

   


**Loss Function**

Para o algoritmo do Perceptron, usamos a seguinte função de perda (*loss function*):

$$L(x_i, y_i) = y_i \cdot (\theta \cdot x_i) \le 0$$

Vimos que $\theta \cdot x_i + \theta_0$ é a classificação em si, e $y_i$ é a label de $x_i$.
Nesse problema, $y_i$ pode ser $1$ ou $-1$.

- Se $\theta \cdot x_i + \theta_0$ for menor que zero e a label for $-1$, o resultado é positivo.
- Se $\theta \cdot x_i + \theta_0$ for menor que zero e a label for $1$, o resultado é negativo.

Assim, a expressão $y_i(\theta \cdot x_i + \theta_0)$ é negativa quando a classificação está errada e positiva quando está correta.
(classificação binária).


*Como atualizar o vetor de parâmetros*



 - Nenhum classificador de fomato: <br>
 $\vec{\theta}$ . $\vec{x}$ = $0$ é capaz de separar perfeitamente os dados, pois ele necessáriamente pela origem(0) <br>

 - Por isso adicionamos um Bias: ${\theta_0}$

    - $\vec{\theta}$ . $\vec{x}$ + ${\theta_0}$ = $0$
