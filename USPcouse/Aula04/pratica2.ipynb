```python
import pandas as pd
```

```python
df = pd.read_csv(
    'datasets_treino_pandas.csv'
)

df[:5]
```

```python
#function date standardization 

def standardizesDates(dataframe, column_name):
    dataframe[column_name] = pd.to_datetime(dataframe[column_name], format='mixed', dayfirst=True);
    return dataframe
```

```python
df = standardizesDates(df,'Data_Pedido')

df['Data_Pedido']
```

```python
df['Desconto'] = df['Desconto'].fillna('Desconto não aplicado')
df['Lucro'] = df['Lucro'].replace('Sem lucro', '0')
df.isna().sum()
```

```python
df['Lucro'] = pd.to_numeric(df['Lucro'], errors='coerce')
df['Valor_Venda'] = pd.to_numeric(df['Valor_Venda'], errors='coerce')

df['Custo'] = df['Valor_Venda'] - df['Lucro'];
print(df[['Custo', 'Valor_Venda', 'Lucro']]);

```

```python
#Sinalizador de rentabilidade

df['Alta_Rentabilidade'] = df['Lucro'] > 200
df['Alta_Rentabilidade']
```

```python
MaioresLucros = df.groupby('Produto')['Lucro'].sum().nlargest(5)
MaioresLucros
```

```python
#Margens negarivas

df_margens_negativas = df.query('Lucro < 0')
df_margens_negativas
```

```python
#Filtragem complexa

df_filtrado = df.query("Segmento == 'Consumidor' and Regiao == 'Sudeste' and Categoria == 'Vestuário'")
df_filtrado
```

```python
# Tabelas dinamicas
df['Categoria'] = df['Categoria'].str.strip()
pivot_table = pd.pivot_table(
    df,
    index='Categoria',
    columns='Regiao',
    values='Quantidade',
    aggfunc='sum',
    fill_value=0
)

pivot_table
```

```python
#Melhor Cliente
df_cliente = df.groupby(['ID_Cliente', 'Segmento'])['Valor_Venda'].sum().sort_values(ascending=False).reset_index()

df_cliente.iloc[0]
```

```python
#Produtos unicos

produto_unicos = df['Produto'].value_counts() == 1
df_unicos = df[df['Produto'].isin(produto_unicos)]
df_unicos
```
