Considere um array bidimencional no qual cada vetor de uma dimenção possui 3 pontuações que um jogador recebeu em 3 provas distintas. Faça uma função que dado esse array, retorne a pova com a maior soma  de pontuações.

```python
import numpy as np
```

```python
def EntMaiorSom(arrBi):
    pont = np.array(arrBi)
    sm = np.sum(pont, axis=0)
    return np.max(sm);
```

```python
values = ([6.5, 7.8, 5.4], [9.5, 8.8, 9.1], [4.5, 7.3, 5.9])

print(EntMaiorSom(values))
```
