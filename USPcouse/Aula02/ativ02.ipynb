*2) Voce está organizando uma lista de alunos por turmas em uma escola. Cada turma contém vários alunos, e cada aluno deve conter um nome, idade e id.*

mesmo exercicio da primera aula porém deve ser resolvido com POO.

a) Criar as turmas: Crie duas turmas, cada ma contendo uma lista de dicionários representado os alunos. Cada dicionário de ter as keys "nome" e "idade".

b) Adicionar um novo aluno: Adicione um nome aluno em uma das turmas.

c) Remover um aluno: Remova um aluno de uma das turmas.

d) Exibir os alunos de uma turma.


```python

class Turma:
    def __init__(self, nome):
        self.nome = nome;
        self.alunos = []
    
    def criarAluno(self, nome, idade):
        self.alunos.append({'id':len(self.alunos),'nome':nome,'idade':idade});
    def removerAluno(self, id):
        for i in range(len(self.alunos)):
            if(self.alunos[i]['id'] == id):
                self.alunos.pop(i);
    def exibirAlunos(self):
        for aluno in self.alunos:
            print(f'Turma: {self.nome}, Id: {aluno['id']}, Nome: {aluno['nome']}, Idade: {aluno['idade']}');




```

```python
turma1 = Turma('A');

turma1.criarAluno('José', 22);
turma1.exibirAlunos();

turma2 = Turma('B');

turma2.criarAluno('Guilherme', 19);

turma2.criarAluno('Mario', 19);
turma2.exibirAlunos();
turma2.removerAluno(1)
turma2.exibirAlunos();
```
