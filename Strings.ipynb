## tabela do ReGex
Sequência Especial 	Significado 	                                                                        Exemplo
\d 	                Combina qualquer personagem de dígito (0-9)                                             "123" corresponde a "\d\d\d"
\D 	                Corresponde a qualquer caractere não-dígito 	                                        "olá" corresponde a "\D\D\D\D"
\w 	                Corresponde a qualquer caractere de palavra (a-z, A-Z, 0-9 e _) 	                    "hello_world" corresponde a "\w\w\w\w\w\w\w\w\w\w\w"
\W 	                Combina qualquer caractere não-palavra 	"@#$%" partidas                                 "\W\W\W"
\s 	                Corresponde a qualquer caractere de espaço em branco (espaço, guia, newline, etc.) 	    "olá mundo" corresponde a "\w\w\w\w\s\w\w\w\w"
\S 	                Corresponde a qualquer personagem não-espaço em branco 	"hello_world" corresponde a     "\S\S\S\S\S\S\S\S\S"
\b 	                Corresponde ao limite entre um caractere de palavra e um caractere não-palavra 	        "gato" combina "\bcat\b" em "O gato sentou-se no tapete"
\B 	                Corresponde a qualquer posição que não seja um limite de palavras 	"gato" combina      "\Bcat\B" em "categoria", mas não em "O gato sentou-se no tapete"

```python
n = 'the phone is 1234567890'

partter = r'\d\d\d\d\d\d\d\d\d\d'

match = re.search(partter, n)

if match:
    print(f'Numero encontrado {match.group()}')
else:
    print('no match')
```

```python
h = 'Hello, World'
partter = r'\W'
match = re.findall(partter, h)

print(f'{match}')
```

```python
m = "michael jackson was a singer and known as the 'king of pop'"
result = re.findall('as',m)
print(result)
```

```python
split_arr = re.split(r"\s",m)
print(split_arr)
```

```python
k = r'king of pop'
l = 'legend'
new_string = re.sub(k,l,m,flags=re.IGNORECASE)
print(new_string)
```
