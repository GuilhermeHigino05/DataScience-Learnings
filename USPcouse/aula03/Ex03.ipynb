Gere um array A contendo todos o numeros impares entre 0 e 100. A partir de A, crie outro array que contenha a soma cumulativa dos elementos de A que sao divisívies por 3 e 5 simultaneamente

```python
import numpy as np
```

```python
arr = np.arange(1, 100, 2)
print(arr)
```

```python
arr2 = arr[(arr %3 == 0) & (arr %5 == 0)].cumsum()
print(arr2)
```
