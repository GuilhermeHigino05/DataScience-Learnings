Faça uma função que dados dois vetores (valores e pesos), calcule a média ponderada de valores com seus respectivos pesos.
Obs: não é permtido usar a função np.avarage do NumPy


```python
import numpy as np
```

```python
def CalcMedPon(val,pes):
    sm = np.sum(val * pes);
    return sm / (np.sum(pes))
```

```python
val = np.array([3, 0, 0, 0, 0, 10, 7, 0, 9])
pes = np.array([1, 0.2, 1, 0.2, 1, 0.2, 1, 0.2, 5.4])
arr = CalcMedPon(val,pes)
print(arr)
```
