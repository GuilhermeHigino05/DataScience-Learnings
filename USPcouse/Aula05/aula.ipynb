**Matiplotlib**

*O que é o matplotlib*
É uma biblioteca de gráficos no python que permite fazer a plotagem de gráficos

*Figure*: figures são o corpo da sua plotagem

*Axes*: Dentro das figures tem os Axes que são o 'Conteúdo' de uma figure

##Pegar a imagem da aula depois


```python
import matplotlib.pyplot as plt
import numpy as np


fig,ax = plt.subplots(figsize=(8, 4));
ax.plot(['Janeiro', 'Fevereiro', "Março", "Abril"], [48, 20, 12, 30]);

ax.set_title('Sales Over Time');
ax.set_xlabel('Month');
ax.set_ylabel('Revenue');
plt.show
```

**Plotagem**

 - A plotagem e a inserção de valores no axes.
 - É comum e simples ignorar processos de criar figures e axes e apenas fazer a plotagem direto, porém o controle sobre o gráfico é perdido .

```python
#Plotagem
#Cria um gráfico comum de linha 
x = np.linspace(0, 10, 100)
y = np.sin(x)#seno de x

plt.plot(x,y);
plt.show()
```

```python
#Scatterplot
#Desenha pontos em um plano

x = np.random.rand(100)
y = np.random.rand(100)

plt.scatter(x,y)
```

```python
#Gráfico de barras padrão

classes = ['Guilherme', 'Murilo', 'Joao'];
notes = [10.0, 8.2, 6.5];

plt.bar(classes, notes)
```

```python
#Histograma

#Faz um gráfico de barras com a frequência de um intervalo

notas = 10 * np.random.rand(100);
xticks = np.arange(0, 11, 1);
yticks = np.arange(0, 25, 2);
plt.hist(notas, bins=10)
plt.xticks(xticks)
plt.yticks(yticks)
plt.title('Histograma de Notas')
```

```python
#Heatmap de uma matrz de dados

pixels = [[0.0, 0.3, 1.0, 0.4, 0.8], [0.5, 1.0, 0.0, 0.9, 0.2], [1.0, 0.0, 0.5, 0.1, 0.6], [0.2, 0.8, 0.3, 1.0, 0.4], [0.9, 0.1, 0.7, 0.5, 0.0]];
plt.imshow(pixels)
```

**CUSTOMIZAÇÃO**

*Legendas*
 - Podemos colocar legendas com o atrbuto label na hora de plotar

```python
fig, ax = plt.subplots();

x = np.linspace(0, 10, 100);
y1 = np.sin(x);
y2 = np.cos(x);

ax.plot(x, y1, label='Seno de x');
ax.plot(x, y2, label='Cosseno de x');
ax.legend()
```

```python
#Escalas

x = np.linspace(0, 10, 100);
y1 = np.exp(x);

plt.plot(x, y1, c='red');
plt.yscale('log');
```

```python
#Cores por ponto

fig, ax = plt.subplots();
x = np.random.rand(500) - 0.5
y = np.random.rand(500) - 0.5

cores = np.sqrt(x**2 + y**2);  #formula da magnitude de um vetor

pontos = ax.scatter(x, y, c=cores, cmap='rainbow')
fig.colorbar(pontos)
```
