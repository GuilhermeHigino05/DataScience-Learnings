Enunciado: Desenvolva um programa que selecione um numero aleatório entre 0 e 100. O usuário deve tentar adivinhar esse número. Ao final, o programa exibirá a sequência de palpites do usuário e o total de tentativas feitas até acertar o número correto.

```python
import random as rd

valor = rd.randint(0, 100);
```

```python
tentativa = []
acertou = False
while not acertou:
    chute = int(input('Write a number between 0 and 100: '));
    tentativa.append(chute)
    if chute == valor:
        acertou = True
    elif chute > valor:
        print('Write a smaller number')
        continue
    else:
        print('Write a bigger number');
        continue
```

```python
print(f'Congratilations you hit in {len(tentativa)} attempts, attempts: {tentativa}')
```
