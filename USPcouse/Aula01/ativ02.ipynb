Enunciado: Recriar a pirâmide do mario. O tamanho da pirâmide será decidida pelo usuário

```python
while True:
    tam = int(input('Write pyramid size'));
    if(tam <= 0):
        print('The pyramid size must be greater than 0');
        continue;
    for i in range(tam):
        for _ in range(tam - (i+1)):
            print(' ', end='');
        for _ in range(i+1):
            print('#', end='');
        print();
    break;
```

```python
while True:
    tam = int(input('Write pyramid size'));
    if(tam <= 0):
        print('The pyramid size must be greater than 0');
        continue;
    for i in range(tam):
        print(' ' * (tam -(i+1)) + '#' * (i+1));
    break;
```
