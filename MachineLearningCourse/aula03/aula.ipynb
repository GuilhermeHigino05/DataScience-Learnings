**Regressão Linear ao Polínomio**

   *Regressão x Classificação*
 A regressão tenta encontrar a melhor função de reta que melhor representa os dados enquanto a classificação tenta separar os dados de forma linear

 

O modelo de regressão linear segue: <br>
 $ Y = aX+b $   Função de reta, onde 'a' é o coeficiente angular e 'b' é o coeficiente linear. <br>

 Para o exemplo do gráfico temos <br>

 $\hat{Y} = 5X + 50$ <br>

 Portanto, para um novo valor de X podemos Prever com: 

 $\hat{Y} = 5*110 + 50 = 600mil$ <br>

Loss Function: MSE (Mean Squared Error) <br>
