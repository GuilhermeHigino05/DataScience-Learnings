**Estatística para numpy**

_Estatisticas descritivas_

- Medidas de tendencias cetrais
  média, moda, mediana e percentil -> Percentil é uma medida que divide um conjunto de dados em 100 pates iguais.
- Valores limites
- Desvio padrão -> O desvio padrão é uma medida de quanto os valores variam em relação a média | se os desvio padrão for baixo os valores estão mais próximos da média se for alto eles estou espalhados. 
- Somatório -> soma e soma acumulativa


```python
arr = np.linspace(0, 20, 21);
arr
```

```python
# média

arr.mean()
```

```python
arr.max()
```

```python
arr.min()
```

```python
# indice onde o valor máximo foi encontrado e onde o valoor minímo foi encontrado
print(arr.argmax())

print(arr.argmin())
```

```python
# desvio padrão 

print(arr.std())
```

```python
arr.sum()
```

```python
arr.cumsum()
```

```python
#Exercicio: Selecione os valores que se distanciam da média até o desvio padrão 

bool_idx = (arr > arr.mean() - arr.std()) & (arr < arr.mean() + arr.std())

print(arr[bool_idx])
```

**Funções universais do Numpy**

São funções vetorizadas que podem operar de maneira eficientes arrays N-dimencioais (ndArrays) elemento a elemento, são escritas em C.

*uFuncs*


```python
#mediana

print(np.median(arr))
```

```python
print(np.percentile(arr, 45))
```

```python
#seno
arr2 = np.array([0, np.pi /2])

np.sin(arr2)
```

```python
#Maximum -> Valor maximo de cada posição

arr_1 = np.array([3, 8, 32, 2, 54, 17])
arr_2 = np.array([ 9, 11, 63, 22, 0, 41])

np.maximum(arr_1, arr_2)
```

```python
#Valores pseudoaleatórios

arr = np.random.uniform(0, 10, 6)
print(arr)
print()
print(np.modf(arr))
```

**Álgebra Linear** 

 - Algebra linear é uma área da matemática que se foca no estudo de vetores, espaços vetoriais e transfomaçõe lineares.

 - Entender e manipular vetores e matrizes são fundamentais para análise e manipulação de dados.

 - Ndarrays 
 -Podemos criar vetores e matrizes través de um objeto ndarray
 -O numpy oferece múltipas feramentas para manipulação de vetores e matrizes
 Ex: 
    -Adição, subtração
    -Multiplicação por escalar
    -Multiplicação de matrizes
    -Cálculo de determinante
    -Transposição
    -Inverção de matriz


```python
#Outra forma de criar matrizes indentidade 

mtx = np.eye(3) + [1, 1.5, 3]
print(mtx)
```

```python
vet = np.array([1, 2, 3])
vet
```

```python
#Multiplicação de matrizes: .dot e @
mtx.dot(vet)
```

```python
mtx @ vet
```

```python
#Determinante
print(np.linalg.det(mtx))
```

```python
#Transposição de uma matriz

mtx.T
```

```python
#Matriz inversa

np.linalg.inv(mtx)
```
