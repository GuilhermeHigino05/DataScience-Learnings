*1) Faça uma função que dado um vetor de inteiros retorne o maior inteiro do valor*

```python
vetor = []

for i in range(10):
    num = int(input('Preencha o vetor'));
    vetor.append(num);

def verificaMaior(vet):
    if(len(vet) == 0):
        return None;
    vet.sort();
    return vet[-1];

print(vetor)
maior = verificaMaior(vetor);
print(f'O maior elemento do vetor é {maior}')
```
