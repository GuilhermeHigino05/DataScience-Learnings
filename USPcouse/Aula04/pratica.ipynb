```python
import pandas as pd
```

```python
df = pd.read_excel(
    'datasets_treino_pandas.xlsx',
    sheet_name='Dados_Brutos'
)
```

```python
#.dropna() remove qualquer linha que tenha pelo menos um valor NaN

#.isna() mapeia os valores nulos e retorna True para cada valor nulo

print(f'Linhas Vazias \n{df.isna().sum()}')

df = df.fillna(0.0)# Preenche todos os valores nulos com 0.0
```

```python
# Tratamento de linas duplicadas

print(f'Quantidade de linhas duplicadas {df.duplicated().sum()}')

df = df.drop_duplicates()
```

```python
#Formatando as strings 

df['Categoria'] = df['Categoria'].str.strip()

print(df['Categoria'][0: 10])
```

```python
#transformando datas

df['Data_Pedido'] = pd.to_datetime(df['Data_Pedido'], errors='coerce')
```

**Perguntas a serem respondidas**

- Qual a Região que gerou maior valor total de vendas?  
- Qual categoria de produtos é mais lucrativa em média?

```python
new_df = df.groupby('Regiao')['Valor_Venda'].sum()

print(f'A região que mais valor total de vendas foi o {new_df.idxmax()}')
```

```python
most_profitable_product_on_avarage = df.groupby('Categoria')['Lucro'].mean()

print(most_profitable_product_on_avarage.idxmax())
```

```python
#Criando linha do tempo de fatramento


df['Linha_Do_Tempo'] = df['Data_Pedido'].dt.to_period('M')
df['Lucro_por_mes'] = df.groupby('Linha_Do_Tempo')['Lucro'].transform('sum')


print(df[['Linha_Do_Tempo', 'Lucro_por_mes']])
```
