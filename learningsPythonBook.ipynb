*Functions and Objects method calls*


```python
#Você chama fucões usando parênteses e passando os argumentos como parâmetro
#ex:
def soma(a,b):
    c = a+b
    return c

g = soma(10,5)
g
```

```python
#Quando criamos funcões e  essas funções recebem objetos como parâmentro esses objetos não são copiando e sim é criado uma referência(ponteiro) para o objeto original
#ex:

def append_element(some_list, element):
    some_list.append(element)
data = [1, 2, 3]
append_element(data, 4)
data
```
