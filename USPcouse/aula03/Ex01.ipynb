Faça uma função que receba um array como entrada e remova o primeiro e o último elemento do array e retorne um novo array contendo os elementos restantes

```python
import numpy as np
```

```python

def RemoveNums(arr):
    new_arr = np.delete(arr,[0, -1])
    return new_arr
```

```python
arr = np.random.uniform(0, 10, 8)
print(arr)

arr_new = RemoveNums(arr)

print(arr_new)
```
