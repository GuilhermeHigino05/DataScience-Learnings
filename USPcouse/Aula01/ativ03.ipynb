Enumciado:Você está organizando uma lista de alunos por turmas em uma escola. Cada turma contém vários alunos, e cada aluno é representado por um dicionário com informações sobre seu nome e idade.

a) Criar as turmas: Crie duas turmas, cada ma contendo uma lista de dicionários representado os alunos. Cada dicionário de ter as keys "nome" e "idade".

b) Adicionar um novo aluno: Adicione um nome aluno em uma das turmas.

c) Remover um aluno: Remova um aluno de uma das turmas.

d) Exibir os alunos de uma turma.
